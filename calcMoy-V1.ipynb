# Calcul des moyennes des UE

```python
# import numpy as np
import matplotlib.pyplot as plt 

def calculU1():
    matiere = [f"R{i}" for i in range(101, 116)] + [
        f"SAE{i}" for i in range(11, 17)  ]
    note = []

    coeff1 = [10, 10, 7, 7, 0, 5, 0, 6, 0, 5, 4, 2, 5, 5, 0, 20, 20, 0, 0, 0, 7]
    coeff2 = [4, 0, 2, 8, 6, 0, 0, 0, 0, 5, 5, 2, 9, 9, 3, 0, 0, 29, 0, 0, 7]
    coeff3 = [4, 0, 2, 0, 0, 5, 15, 6, 4, 5, 5, 2, 0, 0, 3, 0, 0, 0, 20, 20, 7]

    for i in range(101, 116):
        note.append(float(input(f"Quelle est votre moyenne R{i} ? ")))
    for i in range(11, 17):
        note.append(float(input(f"Quelle est votre moyenne SAE{i} ? ")))

    dicnote1 = {cle: valeur for cle, valeur in zip(matiere, note)}
    diccoeff1 = {cle: valeur for cle, valeur in zip(matiere, coeff1)}

    somme = 0
    somme_coeff = 0
    for m in dicnote1:
        somme += dicnote1[m] * diccoeff1[m]
        somme_coeff += diccoeff1[m]
    moy1 = somme / somme_coeff
    print("Votre moyenne pour UE1 est", round(moy1, 2))

    diccoeff2 = {cle: valeur for cle, valeur in zip(matiere, coeff2)}
    somme = 0
    somme_coeff = 0
    for m in dicnote1:
        somme += dicnote1[m] * diccoeff2[m]
        somme_coeff += diccoeff2[m]
    moy2 = somme / somme_coeff
    print("Votre moyenne pour UE2 est", round(moy2, 2))

    diccoeff3 = {cle: valeur for cle, valeur in zip(matiere, coeff3)}
    somme = 0
    somme_coeff = 0
    for m in dicnote1:
        somme += dicnote1[m] * diccoeff3[m]
        somme_coeff += diccoeff3[m]
    moy3 = somme / somme_coeff
    print("Votre moyenne pour UE3 est", round(moy3, 2))

    moyenneGlobal = (moy1 + moy2 + moy3) / 3
    print("Votre moyenne globale est", round(moyenneGlobal, 2))

    if moyenneGlobal >= 10:
        print("Vous passez en BUT 2")
    else:
        print("Vous ne passez pas en deuxième année")

    couleur = ["red" if m < 8 else "orange" if 8 <= m < 10 else "green" for m in [moy1, moy2, moy3]]
    plt.bar(["UE 1", "UE 2", "UE 3"], [moy1, moy2, moy3], color = couleur)
    plt.title("Moyennes des UE")
    plt.ylabel("Notes")
    plt.show()



   
calculU1()
```
