# Calcul des moyennes des UE

```python
# import numpy as np
import matplotlib.pyplot as plt 

def calculU1():
    matiere = [f"R{i}" for i in range(101, 116)] + [
        f"SAE{i}" for i in range(11, 17)  ]
    note = []

    coeff1 = [10, 10, 7, 7, 0, 5, 0, 6, 0, 5, 4, 2, 5, 5, 0, 20, 20, 0, 0, 0, 7]
    coeff2 = [4, 0, 2, 8, 6, 0, 0, 0, 0, 5, 5, 2, 9, 9, 3, 0, 0, 29, 0, 0, 7]
    coeff3 = [4, 0, 2, 0, 0, 5, 15, 6, 4, 5, 5, 2, 0, 0, 3, 0, 0, 0, 20, 20, 7]

    for i in range(101, 116):
        note.append(float(input(f"Quelle est votre moyenne R{i} ? ")))
    for i in range(11, 17):
        note.append(float(input(f"Quelle est votre moyenne SAE{i} ? ")))

    dicnote1 = {cle: valeur for cle, valeur in zip(matiere, note)}
    diccoeff1 = {cle: valeur for cle, valeur in zip(matiere, coeff1)}

    somme = 0
    somme_coeff = 0
    for m in dicnote1:
        somme += dicnote1[m] * diccoeff1[m]
        somme_coeff += diccoeff1[m]
    moy1 = somme / somme_coeff
    print("Votre moyenne pour UE1 est", round(moy1, 2))

    diccoeff2 = {cle: valeur for cle, valeur in zip(matiere, coeff2)}
    somme = 0
    somme_coeff = 0
    for m in dicnote1:
        somme += dicnote1[m] * diccoeff2[m]
        somme_coeff += diccoeff2[m]
    moy2 = somme / somme_coeff
    print("Votre moyenne pour UE2 est", round(moy2, 2))

    diccoeff3 = {cle: valeur for cle, valeur in zip(matiere, coeff3)}
    somme = 0
    somme_coeff = 0
    for m in dicnote1:
        somme += dicnote1[m] * diccoeff3[m]
        somme_coeff += diccoeff3[m]
    moy3 = somme / somme_coeff
    print("Votre moyenne pour UE3 est", round(moy3, 2))

    moyenneGlobal = (moy1 + moy2 + moy3) / 3
    print("Votre moyenne globale est", round(moyenneGlobal, 2))

    if moyenneGlobal >= 10:
        print("Vous passez en BUT 2")
    else:
        print("Vous ne passez pas en deuxième année")

    couleur = ["red" if m < 8 else "orange" if 8 <= m < 10 else "green" for m in [moy1, moy2, moy3]]
    plt.bar(["UE 1", "UE 2", "UE 3"], [moy1, moy2, moy3], color = couleur)
    plt.title("Moyennes des UE")
    plt.ylabel("Notes")
    plt.show()



   
calculU1()
```

```python
#importation des modules neccesaire 
from tkinter import *
from tkinter import ttk
def calcul():
    
#création de la fenetre
fenetre = Tk() 
# Le cadre (frm) va contenir notre grille de matières alignées
frm=ttk.Frame(fenetre)

#le titre de la fenetre 
fenetre.title("Moyenne du semestre 1") 

# création d'un premier texte en haut de fenetre 
Titre = Label(fenetre,text = "Semestre 1",padx=20,pady=20,font=("Arial",50)) 
soustitre = Label(fenetre,text = "Moyenne",padx=20,pady=20,font=("Arial",25)) 

# Boucle automatique pour créer les Label et les Entry (de R101 à R115)
for i in range(101, 116):
  # i prend successivement les valeurs de 101 à 115
  lbl_r101 = ttk.Label(frm, text=f"R{i} : ", font=("Arial", 14))
  donnee_r101 = ttk.Entry(frm, width=30)
  
  # On place chaque élément sur sa propre ligne (row) grâce à l'indice i
  lbl_r101.grid(row=i, column=0, sticky="e", pady=2)
  donnee_r101.grid(row=i, column=1, pady=2) 

# Boucle automatique pour crée les labels et les entry (de SAE 11 à 16)
for i in range(11,17):
  lbl_sae11 = ttk.Label(frm,text=f"SAE{i} : " , font=("Arial", 14))
  donnee_SAE11 = ttk.Entry(frm , width = 30)
 
  lbl_sae11.grid(row=i, column=0, sticky="e", pady=2)
  donnee_SAE11.grid(row=i, column=1, pady=2) 

#affichage des widgets
Titre.pack()
soustitre.pack()
# On affiche le cadre global à gauche à l'extérieur de la boucle
frm.pack(pady=10, padx=20, anchor="w")

#démarrage de la fenetre
fenetre.mainloop()
```
