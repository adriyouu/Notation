# Calcul des moyennes des UE

Demandez à l'utilisateur ses notes dans les differrents UE

```python
import numpy as np      #importation du module numpy 


def calculU1(): #création de la fontion calculU1
    note = []
    noteU2 = [] # initialisation des variables pour les notes en liste vide
    noteU3 = []

    coeff1 = [10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7]
    coeff2 = [4,0,2,8,6,0,0,0,0,5,5,2,9,9,3,0,0,29,0,0,7]    #initialisation des variables pour les coefficients sous forme de liste 
    coeff3 = [4,0,2,0,0,5,15,6,4,5,5,2,0,0,3,0,0,0,20,20,7]

    matiere = [f"R{i}" for i in range(101,116)] + [f"SAE{i}" for i in range(11,17)] #initialisation de la variable matiere 

    for i in range(101, 116):
        note.append(float(input(f"Quelle est votre moyenne R{i} pour UE1 ? "))) #demande à l'utilisateur ses notes pour les ressources de UE1

    for i in range(11, 17):
        note.append(float(input(f"Quelle est votre moyenne SAE{i} ? ")))  #demande à l'utilisateur ses notes pour les SAE de UE1

    dicnote1 = {cle: valeur for cle, valeur in zip(matiere, note)}    #création d'un dictionnaire pour les notes 
    diccoeff1 = {cle: valeur for cle, valeur in zip(matiere, coeff1)} #création d'un dictionnaire pour les coefficients 

    somme = 0  
    somme_coeff = 0

    for matiere in dicnote1:
        somme += dicnote1[matiere] * diccoeff1[matiere]     # calcul de la somme des notes à partir du dictionnaire 
        somme_coeff += diccoeff1[matiere]                   # calcul de la somme des coefficients à partir du dictionnaires 

    moy1 = somme / somme_coeff #calcul de la moyenne de UE1

    print("Votre moyenne pour UE1 est", round(moy1, 2)) #affiche la valeur de la moyenne avec deux chiffres apres la virgule

    for i in range(101, 116):
        noteU2.append(float(input(f"Quelle est votre moyenne R{i} pour UE2 ? ")))

    for i in range(11, 17):
        noteU2.append(float(input(f"Quelle est votre moyenne SAE{i} ? ")))

    dicnote2 = {cle: valeur for cle, valeur in zip(matiere, noteU2)}
    diccoeff2 = {cle: valeur for cle, valeur in zip(matiere, coeff2)}

    somme = 0
    somme_coeff = 0

    for matiere in dicnote2:
        somme += dicnote2[matiere] * diccoeff2[matiere]
        somme_coeff += diccoeff2[matiere]

    moy2 = somme / somme_coeff

    print("Votre moyenne pour UE2 est", round(moy2, 2))

    for i in range(101, 116):
        noteU3.append(float(input(f"Quelle est votre moyenne R{i} pour UE3 ? ")))

    for i in range(11, 17):
        noteU3.append(float(input(f"Quelle est votre moyenne SAE{i} ? ")))

    dicnote3 = {cle: valeur for cle, valeur in zip(matiere, noteU3)}
    diccoeff3 = {cle: valeur for cle, valeur in zip(matiere, coeff3)}

    somme = 0
    somme_coeff = 0

    for matiere in dicnote3:
        somme += dicnote3[matiere] * diccoeff3[matiere]
        somme_coeff += diccoeff3[matiere]

    moy3 = somme / somme_coeff

    print("Votre moyenne pour UE3 est", round(moy3, 2)) 

    moyenneGlobal = (moy1 + moy2 + moy3) / 3

    print("Votre moyenne globale est", round(moyenneGlobal, 2)) #affiche la moyenne global du semestre 1

    if moyenneGlobal >= 10:  #compare la moyenne du semestre à 10 pour savoir si on passe l'année 
        print("Vous passez en BUT 2")
    else: 
        print("Vous ne passez pas en deuxième année")


calculU1()
```
